# Retrieval-Augmented Generation (RAG) Tutorial

This notebook loads the C++ notes from `CPP.txt`, splits the text into chunks, creates embeddings, stores them in a FAISS vector database, and performs similarity searches.

```python
# Data loading step
from langchain_community.document_loaders import TextLoader

loader = TextLoader('CPP.txt', encoding='utf-8')
cpp_data = loader.load()
cpp_data
```

```python
# Text splitter step
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20,
)

final_documents = text_splitter.split_documents(cpp_data)
len(final_documents)
```

```python
# Embedding step
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name='sentence-transformers/all-MiniLM-L6-v2'
)

print(embeddings.model_name)
```

```python
# Build the FAISS vector store
from langchain_community.vectorstores import FAISS

db = FAISS.from_documents(final_documents, embeddings)
db
```

```python
# Similarity search
query = input('Enter your query: ')
docs = db.similarity_search(query)

for document in docs:
    print(document.page_content)
    print('\n')
```
